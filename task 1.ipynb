```python
# Store values in variables
celsius = 80.8

# Formula to convert Celsius to Fahrenheit
fahrenheit = (celsius * 9 / 5) + 32

# Display the output
print(f"{celsius}°C is equal to {fahrenheit}°F")
```

```python
# Define a function with an input parameter
def get_letter_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"

# Call the function with different values
student_score = 85
grade = get_letter_grade(student_score)

print(f"Score: {student_score} | Grade: {grade}")
```

```python
def analyze_numbers(numbers):
    """Calculates sum, average, and finds even numbers in a list."""
    total_sum = 0
    even_numbers = []

    # Loop through the list to accumulate sum and filter evens
    for num in numbers:
        total_sum += num
        if num % 2 == 0:
            even_numbers.append(num)

    average = total_sum / len(numbers)

    # Return multiple results
    return total_sum, average, even_numbers

# Main program setup
sample_data = [12, 7, 19, 24, 5, 10, 16]

# Execute function
total, avg, evens = analyze_numbers(sample_data)

# Print detailed summary
print("--- Number Analysis Report ---")
print(f"Original Data: {sample_data}")
print(f"Total Sum:     {total}")
print(f"Average:       {avg:.2f}")
print(f"Even Numbers:  {evens}")
```
